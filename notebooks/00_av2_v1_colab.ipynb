**Dataset:** CDC, *Mapping Injury, Overdose, and Violence - State*.

**Pergunta:** Saber quais leis estaduais sobre porte e verificação de antecedentes estavam em vigor ajuda a prever a taxa de homicídios do ano seguinte em cada estado americano?

Nesta primeira entrega, apenas abrimos os dados com Pandas e conferimos o conteúdo. A análise fica para a próxima etapa. Fonte: https://data.cdc.gov/d/fpsi-y8tj


```python
import pandas as pd
import requests
from io import StringIO

url = "https://data.cdc.gov/resource/fpsi-y8tj.csv"
parametros = {
    "$where": "intent='All_Homicide' AND period IN ('2019','2020','2021','2022','2023','2024')",
    "$limit": 1000,
}
resposta = requests.get(url, params=parametros, timeout=40)
resposta.raise_for_status()
dados = pd.read_csv(StringIO(resposta.text))
dados.head()
```
