**Bases escolhidas:** CDC (*Mapping Injury, Overdose, and Violence - State*) e RAND (*State Firearm Law Database 6.0*).

**Pergunta:** existe relação entre a quantidade de regras estaduais sobre armas registradas na RAND e a taxa de homicídios nos estados americanos?

**Hipótese inicial:** estados com mais regras registradas podem apresentar taxas de homicídios diferentes dos estados com menos regras. Ainda não sabemos se essa diferença existe nem em qual direção.

Nesta primeira entrega, apenas abrimos as duas bases com Pandas. No projeto, contamos os registros `Implement` e `Modify` vigentes em 31 de dezembro, separados entre `Restrictive` e `Permissive`. A análise da relação entre as bases fica para a próxima etapa. Uma associação entre os dados não prova que as leis causaram uma mudança nos homicídios.

Fontes: [CDC](https://data.cdc.gov/d/fpsi-y8tj) e [RAND](https://www.rand.org/pubs/tools/TLA243-2-v4.html).


```python
import pandas as pd
import requests
from io import StringIO

url_cdc = "https://data.cdc.gov/resource/fpsi-y8tj.csv"
parametros = {
    "$where": "intent='All_Homicide' AND period IN ('2019','2020','2021','2022','2023','2024')",
    "$limit": 1000,
}
resposta = requests.get(url_cdc, params=parametros, timeout=40)
resposta.raise_for_status()
homicidios = pd.read_csv(StringIO(resposta.text))
print("CDC:", homicidios.shape)
homicidios[["name", "period", "rate", "count_sup"]].head()
```
